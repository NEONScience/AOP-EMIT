```python
#construct example DataFrame
ex_df = pd.DataFrame(
    #with this dict() we don't have to put key names in quotes
    dict(
        values=[random.randint(1, 1000)
                #produces a generator that starts at 0 and counts up to 99
                # this lets us run the loop a certain number of times
                for i in range(100)],
        group1=[random.randint(1, 100) for i in range (25)] * 4,
        #pick 4 random integers btw 1 and 100 and repeat thos 4 numbers over and over 25 times
        group2=[random.randint(1, 100) for i in range (4)] * 25)
    
)
ex_df
```

```python
#loop through dataframe by group
for group2, df in ex_df.groupby('group2'):
    print(group2)
    display(df)

print(group2)
```

```python
#set up accumulator list
squared_df_list = []
for group1, df in tqdm(ex_df.groupby('group1')):
    print(f'Group {group1}:')
    squared_values = (
        #sauare dataframe values column
        df['values'] ** 2)
    #add back to df in new column
    df['squared_values'] = squared_values
    #instead of the two lines above, it could just be 
    # df['squared_values']=df['values'] ** 2
    #append new df to list
    squared_df_list.append(df)

#glue the 25 separate lists into one df
pd.concat(squared_df_list)
```

```python
#looping w/ iterrows method

#this is a generator which means we can loop through it
ex_df.iterrows()

#this should give us the index of each row and the row contents as a series for each row
for i, row in ex_df.iterrows():
    print (i)
    print (row)
```

```python
#this should give us the index of each row and the row contents as a series for each row
for i, row in ex_df.iterrows():
    print (f'Row {i}:')
    print (row)
    #add a blank line in btw each
    print('\n')
```

```python
#this should give us the index of each row and the row contents as a series for each row
#formatting prints in this way can help us debug to see different printouts clearly
for i, row in ex_df.iterrows():
    print (f'Row {i}:')
    #\t indents
    print (f"\tValue: {row['values']}")
    print (f"\tGroup1: {row['group1']}")
    print (f"\tGroup2: {row['group2']}")
    print('\n')
```

```python
#create accumulator list
new_row_list = []

#we can use this to have loop run a few times w/o the break
i = 0

for row_no, row in tqdm(ex_df.iterrows()):
    if i > 5:
        break
    #make a dictionary to add to new_row_list
    new_row = dict(
        values = row['values'],
        grp1_plus_grp2 = row['group1'] + row['group2']
    )
    print(new_row)
    new_row_list.append(new_row)
    #this lets us see some progress of the loop w/o seeing the structure of the whole list
    print(f'Number of rows: {len(new_row_list)}\n')
    i += 1

#we get a list of dictionaries, possibly easier to work w/ list of dictionaries than dictionaries of list
new_row_list
```
