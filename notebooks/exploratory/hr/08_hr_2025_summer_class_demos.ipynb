```python
#set up accumulator list
squared_df_list = []
for group1, df in tqdm(ex_df.groupby('group1')):
    print(f'Group {group1}:')
    squared_values = (
        #sauare dataframe values column
        df['values'] ** 2)
    #add back to df in new column
    df['squared_values'] = squared_values
    #instead of the two lines above, it could just be 
    # df['squared_values']=df['values'] ** 2
    #append new df to list
    squared_df_list.append(df)

#glue the 25 separate lists into one df
pd.concat(squared_df_list)
```

## try/except

can't use try by itself, it must have an except.

```python
raise ValueError('Error Message')
```

```python
try:
    #usually we put functions in here and the fxn tries to run
    #if the function can't run, the except will be printed (I think)
    raise ValueError('Error Message')
except:
    print('that was incorrect.')
```

```python
a = 15
a.append(5)
#this throws and attirbute error that int object has no attribute instead
```

```python
a = 15
try:
    a.append(5)
#instead of throwing the attribute error, put a in a list and append to the list instead
#if possible, we want to be specific about the type of error in the except
#this can be tricky if a library has a custom exception
except AttributeError:
    a = [a]
    a.append(5)

a
```

```python
a = 15
try:
    a.append(5)
#instead of throwing the attribute error, put a in a list and append to the list instead
except AttributeError as err:
    #this still lets us see the error message but still moves on with the code 
    print(err)
    a = [a]
    a.append(5)

a
```

## While loops

```python
num = 1
prev = 0
fibonacci = []

#this runs a for loop where instead of stopping on it's own,
#this will keep running while the conditional is true
while num<100:
    fibonacci.append(num)
    old_num = num
    num = prev + num
    prev = old_num
    
fibonacci
```
