# Seeing the Forest as well as the Trees - Confirming Forest Measurements from Space

## Project Description
Mapping can help prioritize interventions to improve forest health and mitigate
water stress reducing the risk of future high-intensity wildfires, especially 
with the added pressures of climate change. [1] One of the questions driving 
this project is the scale at which forest managers evaluate data and are able 
to effectively diagnose forest health. Ideally, forest managers could use 
regional data from satellites to identify areas under stress and then narrow 
down management plans to smaller plots within that region.

Forest health has been measured in a number of ways including rates of tree 
mortality and the risk of fire damage due to stands of dead trees [2]. In this 
project, we used data from the NEON Soaproot Saddle Site [3] in the Sierra 
National Forest in California (shown in figure 1) to evaluate areas that burned 
during 2020-21 and those that remained unburned to calculate canopy water 
content (CWC). 

The calculation of canopy water content (CWC) from NEON data at 1 meter 
resolution was compared with EMIT data from the International Space Station 
at 60 meter resolution. [4] to see if there was agreement.
