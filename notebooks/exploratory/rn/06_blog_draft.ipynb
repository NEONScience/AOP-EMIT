# Seeing the Forest as well as the Trees - Confirming Forest Measurements from Space

## Project Description
Questions we investigated included "Can low resolution data from space agree 
with high resolution data from an aerial survey?" and "Can maps from space
effectively diagnose forest health?" Ideally, forest managers could use regional 
data from satellites to identify areas under stress and then narrow down 
management plans to smaller plots within that region. Mapping can help prioritize 
interventions to improve forest health and mitigate water stress reducing the 
risk of future high-intensity wildfires, especially with the added pressures of 
climate change. [1] 

Forest health has been measured in a number of ways including rates of tree 
mortality and the risk of fire damage due to stands of dead trees [2]. In this 
project, we used data from the NEON Soaproot Saddle Site [3] in the Sierra 
National Forest in California (shown in figure 1) to evaluate areas that burned 
during 2020-21 and those that remained unburned to calculate canopy water 
content (CWC). 

The calculation of canopy water content (CWC) compared high resolution data from 
the National Ecological Observatory Network (NEON) at 1 meter resolution with 
low resolution data from the International Space Station Earth Surface Mineral 
Dust Source Investigation (EMIT) instrument at 60 meter resolution. [4] to see 
if there was agreement.

## Project Methods
Both NEON and EMIT instruments passively collect reflected solar radiation in 
285 to 426 wavelengths. This is called Hyperspectral. If you are unfamiliar 
with Hyperspectral Remote Sensing, refer to this [NEON tutorial](https://www.neonscience.org/resources/learning-hub/tutorials/hyper-spec-intro) 
before proceeding.

Two 1km by 1 km areas were identified for comparison within the NEON SOAP site. 
These areas are called tiles. The green and orange tiles shown in figure 2 are 
adjacent to each other and have similar environments, but one tile was burned 
by the Fall 2020 Creek Fire while the other was not. After these two NEON tiles 
were chosen, a granule from EMIT was identified that covered both of these areas. 
A granule represents a specific spatiotemporal chunk of data. It's tied to a 
particular acquisition event (time) and covers a defined geographic area.

Measurements of reflected solar radiation by imaging spectrometers can quantify 
water in different states (solid, liquid, gas) and using the Beer-Lambert law 
to calculate canopy water content. [5]

## Discussion of Project Findings
We expected CWC calculations using the high-resolution NEON data from the SOAP 
site for both burned and unburned tiles to be similar to the calculations from 
the wider area surveyed by EMIT. After calculating CWC with NEON and EMIT 
reflectance data for the burned and unburned tiles and plotting the CWC, visually 
the plots show general agreement on areas with very high or very low CWC. The 
different resolutions between EMIT and NEON are also very obvious. 
(see figure 3).

```python
# --- Configuration ---
# Current working directory of the notebook (used for robust pathing)
current_working_directory = os.getcwd()
image_folder = os.path.join(current_working_directory, 'img')

# List of your image filenames
image_filenames = [
    'EMIT_burned.png', # Replace with your actual first image file name
    'NEON_burned.png', # Replace with your actual second image file name
    'EMIT_unburned.png', # Replace with your actual third image file name
    'NEON_burned.png'  # Replace with your actual fourth image file name
]

# Optional: List of titles for each image (should match the order of image_filenames)
image_titles = [
    "Image 1: Low resolution after burn",
    "Image 2: High resolution after burn",
    "Image 3: Low resolution before burn",
    "Image 4: High resolution before burn"
]

# Optional: Main title for the entire figure
main_figure_title = "Comparative Analysis of SOAP Site Tiles"

# Optional: Overall legend/caption for the entire figure
overall_figure_legend = (
    "Figure 3: This plot displays canopy water content calculations"
    "from the SOAP site, showing the difference in resolution"
    "between EMIT (low) and NEON (high)."
)

# --- Create the plot grid ---
# Create a figure and a 2x2 grid of subplots
# figsize: (width, height) in inches. Adjust as needed.
# We unpack the 'axs' into a 2D array: axs[row, col]
fig, axs = plt.subplots(2, 2, figsize=(14, 12)) # Example figsize, adjust as needed

# Flatten the axs array for easier iteration if preferred, or use 2D indexing
# axs_flat = axs.flatten()

# --- Load and display each image ---
for i, filename in enumerate(image_filenames):
    image_path = os.path.join(image_folder, filename)

    # Determine the subplot position (row, col) in a 2x2 grid
    row = i // 2 # Integer division for row: 0, 0, 1, 1
    col = i % 2  # Modulo for column: 0, 1, 0, 1

    try:
        img = Image.open(image_path)
        print(f"Loading {filename}...")

        # Display the image in the current subplot
        axs[row, col].imshow(img)

        # Hide axes ticks and labels for a cleaner image display
        axs[row, col].axis('off')

        # Add a title to each subplot
        if i < len(image_titles): # Check if a title exists for this image
            axs[row, col].set_title(image_titles[i], fontsize=14, pad=10)

    except FileNotFoundError:
        print(f"Error: Image not found for '{filename}' at {image_path}")
        axs[row, col].set_title(f"Image not found: {filename}")
        axs[row, col].axis('off') # Still hide axes if error
    except Exception as e:
        print(f"An error occurred loading '{filename}': {e}")
        axs[row, col].set_title(f"Error: {filename}")
        axs[row, col].axis('off')

# --- Add overall figure elements (title and legend) ---

# Add a main title for the entire figure
fig.suptitle(main_figure_title, fontsize=18, y=0.98) # 'y' adjusts vertical position

# Add the overall figure legend (caption) below all subplots
fig.text(0.5, 0.02, overall_figure_legend, ha='center', fontsize=12, wrap=True)

# Adjust layout to prevent text/image overlap
# rect=[left, bottom, right, top] - values are normalized from 0 to 1
plt.tight_layout(rect=[0, 0.06, 1, 0.94]) # Adjust bottom and top margins

# Show the combined plot in the notebook
plt.show()

print(f"Current working directory: {current_working_directory}")
```
